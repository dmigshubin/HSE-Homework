```python
from pymongo import MongoClient
import random
import datetime

client = MongoClient("mongodb://root:password@localhost:27017/")
db = client["analytics"]

sessions_col = db["UserSessions"]
events_col = db["EventLogs"]
tickets_col = db["SupportTickets"]
recommendations_col = db["UserRecommendations"]
reviews_col = db["ModerationQueue"]

# Генерация UserSessions
pages = ["/home","/products","/cart","/checkout","/profile","/products/42"]
devices = ["mobile","desktop","tablet"]
actions = ["login","view_product","add_to_cart","checkout","logout"]

sessions = []
for i in range(100):
    start = datetime.datetime.utcnow() - datetime.timedelta(minutes=random.randint(1,500))
    session = {
        "session_id": f"sess_{i}",
        "user_id": f"user_{random.randint(1,20)}",
        "start_time": start,
        "end_time": start + datetime.timedelta(minutes=random.randint(5,60)),
        "pages_visited": random.sample(pages, random.randint(2,5)),
        "device": random.choice(devices),
        "actions": random.sample(actions, random.randint(2,4))
    }
    sessions.append(session)

sessions_col.insert_many(sessions)
print("Inserted sessions:", len(sessions))

# Генерация EventLogs
event_types = ["click","view","purchase","scroll"]
events = []

for i in range(200):
    events.append({
        "event_id": f"evt_{i}",
        "timestamp": datetime.datetime.utcnow(),
        "event_type": random.choice(event_types),
        "details": {
            "page": random.choice(pages)
        }
    })

events_col.insert_many(events)
print("Inserted events:", len(events))

# SupportTickets
statuses = ["open","closed","in_progress"]
issues = ["payment","delivery","login","product"]

tickets = []
for i in range(50):
    created = datetime.datetime.utcnow() - datetime.timedelta(days=random.randint(1,30))
    ticket = {
        "ticket_id": f"ticket_{i}",
        "user_id": f"user_{random.randint(1,20)}",
        "status": random.choice(statuses),
        "issue_type": random.choice(issues),
        "messages": [
            {"sender": "user", "message": "Problem with order", "timestamp": created},
            {"sender": "support", "message": "We are checking", "timestamp": created + datetime.timedelta(hours=1)}
        ],
        "created_at": created,
        "updated_at": created + datetime.timedelta(hours=1)
    }
    tickets.append(ticket)

tickets_col.insert_many(tickets)
print("Inserted tickets:", len(tickets))

# Генерация UserRecommendations
recommendations = []
for i in range(20):
    recommendations.append({
        "user_id": f"user_{i}",
        "recommended_products": [f"prod_{random.randint(1,500)}" for _ in range(3)],
        "last_updated": datetime.datetime.utcnow()
    })

recommendations_col.insert_many(recommendations)
print("Inserted recommendations:", len(recommendations))

# ModerationQueue
reviews = []
for i in range(100):
    reviews.append({
        "review_id": f"rev_{i}",
        "user_id": f"user_{random.randint(1,20)}",
        "product_id": f"prod_{random.randint(1,200)}",
        "review_text": "Great product!",
        "rating": random.randint(1,5),
        "moderation_status": random.choice(["pending","approved","rejected"]),
        "flags": random.sample(["contains_images","spam","offensive"], random.randint(0,2)),
        "submitted_at": datetime.datetime.utcnow()
    })

reviews_col.insert_many(reviews)
print("Inserted reviews:", len(reviews))

# Проверка коллекций
print("Collections in DB:", db.list_collection_names())
print("Sample sessions:", list(sessions_col.find().limit(3)))
```
