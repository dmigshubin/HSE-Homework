```scala
%spark.pyspark
from pyspark.sql import functions as F
users = spark.createDataFrame(
[ ("u1", "Berlin"),
("u2", "Berlin"),
("u3", "Munich"),
("u4", "Hamburg"), ],
["user_id", "city"] )
orders = spark.createDataFrame(
[ ("o1", "u1", "p1", 2, 10.0),
("o2", "u1", "p2", 1, 30.0),
("o3", "u2", "p1", 1, 10.0),
("o4", "u2", "p3", 5, 7.0),
("o5", "u3", "p2", 3, 30.0),
("o6", "u3", "p3", 1, 7.0),
("o7", "u4", "p1", 10, 10.0), ],
["order_id", "user_id", "product_id", "qty", "price"] )
products = spark.createDataFrame(
[ ("p1", "Ring VOLA"),
("p2", "Ring POROG"),
("p3", "Ring TISHINA"), ],
["product_id", "product_name"] )
users.show() 
orders.show() 
products.show()
```

```scala
%spark.pyspark
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```scala
%spark.pyspark
print("USERS:", users.columns)
print("ORDERS:", orders.columns)  
print("PRODUCTS:", products.columns)
```

```scala
%spark.pyspark
orders = orders.withColumn("revenue", F.col("qty") * F.col("price"))
```

```scala
%spark.pyspark
orders.show()
```

```scala
%spark.pyspark
mart = orders.join(users, "user_id", "inner") \
             .join(products, "product_id", "inner")
```

```scala
%spark.pyspark
mart.show()
```

```scala
%spark.pyspark
metrics = mart.groupBy("city", "product_id", "product_name") \
              .agg(
                  F.count("order_id").alias("orders_cnt"),
                  F.sum("qty").alias("qty_sum"),
                  F.sum("revenue").alias("revenue_sum")
              )
```

```scala
%spark.pyspark
metrics.show()
```

```scala
%spark.pyspark
windowSpec = Window.partitionBy("city").orderBy(F.desc("revenue_sum"))
mart_city_top_products = metrics.withColumn("rank", F.rank().over(windowSpec)) \
                                .filter(F.col("rank") <= 2) \
                                .drop("rank")
```

```scala
%spark.pyspark
mart_city_top_products.show()
```

```scala
%spark.pyspark
mart_city_top_products.write.mode("overwrite").parquet("/tmp/sandbox_zeppelin/mart_city_top_products/")
```

```scala
%spark.pyspark
result = spark.read.parquet("/tmp/sandbox_zeppelin/mart_city_top_products/")
result.orderBy("city", F.desc("revenue_sum")).show(truncate=False)
```
